# Latencia y throughput

Mide el tiempo de cómputo de un lote y las muestras procesadas por segundo. El benchmark prepara los datos antes de cronometrar y ejecuta cinco lotes de calentamiento.

```python
import subprocess
import sys
from pathlib import Path

script = Path("scripts/benchmark.py")
if not script.is_file():
    script = Path("../scripts/benchmark.py")

batch_sizes = [1, 8, 32]
runs = 100
result = subprocess.check_output(
    [sys.executable, str(script), "--batch-sizes", *map(str, batch_sizes), "--runs", str(runs)],
    text=True,
)
print(result, end="")
```
